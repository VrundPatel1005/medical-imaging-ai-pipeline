# Segmentation Demo

This notebook demonstrates the same workflow as the command-line scripts: load a scan, load or create a mask, visualize an overlay, and compute metrics.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'scripts'))

from io_utils import binary_mask, load_volume
from metrics import dice, jaccard, mask_stats
from visualize import normalize_for_display
import matplotlib.pyplot as plt
```

```python
image_path = PROJECT_ROOT / 'data/raw/Task09_Spleen/imagesTr/spleen_2.nii.gz'
label_path = PROJECT_ROOT / 'data/raw/Task09_Spleen/labelsTr/spleen_2.nii.gz'

if not image_path.exists() or not label_path.exists():
    print('Download MSD Task09 Spleen first, then rerun this notebook.')
else:
    image = load_volume(image_path)
    mask = binary_mask(load_volume(label_path).data)
    print(mask_stats(mask, image.spacing_zyx))
```

```python
if image_path.exists() and label_path.exists():
    z = image.data.shape[0] // 2
    plt.figure(figsize=(7, 7))
    plt.imshow(normalize_for_display(image.data[z]), cmap='gray')
    plt.imshow(mask[z], cmap='autumn', alpha=0.45)
    plt.axis('off')
    plt.title(f'Segmentation overlay, slice {z}')
```
