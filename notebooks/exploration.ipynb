```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'scripts'))

from io_utils import load_volume
from visualize import normalize_for_display
import matplotlib.pyplot as plt
```

```python
scan_path = PROJECT_ROOT / 'data/raw/Task09_Spleen/imagesTr/spleen_2.nii.gz'
print(scan_path)
print('Update scan_path after downloading the dataset.')
```

```python
if scan_path.exists():
    volume = load_volume(scan_path)
    print(volume.data.shape, volume.spacing_xyz)
    mid = volume.data.shape[0] // 2
    plt.imshow(normalize_for_display(volume.data[mid]), cmap='gray')
    plt.axis('off')
    plt.title(f'Axial slice {mid}')
else:
    print('Dataset not found yet. Follow README download instructions first.')
```
